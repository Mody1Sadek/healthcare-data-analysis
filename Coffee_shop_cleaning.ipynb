```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("dirty_cafe_sales.csv")
df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df) * 100).round(2)
missing_summary = pd.DataFrame({"Missing Values": missing_values,"Missing %": missing_percentage})

missing_summary = missing_summary.sort_values("Missing %",ascending=False)

missing_summary
```

```python
invalid_values = ["ERROR", "UNKNOWN", "None", "unknown", "error", ""]

for column in df.columns:
    count = df[column].astype(str).str.strip().isin(invalid_values).sum()

    print(f"{column}: {count} invalid values")
```

```python
print("Number of duplicated rows:", df.duplicated().sum())
```

```python
clean_df = df.copy()

clean_df.shape
```

```python
clean_df.columns = (clean_df.columns.str.strip().str.lower().str.replace(" ", "_"))

clean_df.columns
```

```python
invalid_values = ["ERROR","UNKNOWN","None","unknown","error",""]

clean_df = clean_df.replace(invalid_values, np.nan)

clean_df.head()
```

```python
text_columns = ["item","payment_method","location"]

for column in text_columns:
    clean_df[column] = (clean_df[column].astype("string").str.strip())
```

```python
numeric_columns = ["quantity","price_per_unit","total_spent"]

for column in numeric_columns:
    clean_df[column] = pd.to_numeric(clean_df[column],errors="coerce")

clean_df[numeric_columns].dtypes
```

```python
clean_df["transaction_date"] = pd.to_datetime(clean_df["transaction_date"],errors="coerce")
```

```python
clean_df.info()
```

```python
clean_df.isnull().sum()
```

```python
payment_mode = clean_df["payment_method"].mode()[0]

clean_df["payment_method"] = (clean_df["payment_method"].fillna(payment_mode))

print("Payment Method mode:", payment_mode)
```

```python
location_mode = clean_df["location"].mode()[0]

clean_df["location"] = (clean_df["location"].fillna(location_mode))

print("Location mode:", location_mode)
```

```python
clean_df.loc[
    clean_df["quantity"].isna() & clean_df["total_spent"].notna() & clean_df["price_per_unit"].notna(),"quantity",
] = (
    clean_df["total_spent"] / clean_df["price_per_unit"]
)

clean_df["quantity"] = clean_df["quantity"].round().astype(int)
```

```python
clean_df.loc[
    clean_df["total_spent"].isna()
    & clean_df["quantity"].notna()
    & clean_df["price_per_unit"].notna(),
    "total_spent",
] = (
    clean_df["quantity"] * clean_df["price_per_unit"]
)
```

```python
clean_df.loc[
    clean_df["price_per_unit"].isna()
    & clean_df["total_spent"].notna()
    & clean_df["quantity"].notna(),
    "price_per_unit",
] = (
    clean_df["total_spent"] / clean_df["quantity"]
)
```

```python
price_to_item = {
    1.0: "Cookie",
    1.5: "Tea",
    2.0: "Coffee",
    5.0: "Salad",
  }

clean_df.loc[
    clean_df["item"].isna()
    & clean_df["price_per_unit"].isin(price_to_item.keys()),
    "item",
] = clean_df.loc[
    clean_df["item"].isna()
    & clean_df["price_per_unit"].isin(price_to_item.keys()),
    "price_per_unit",
].map(
    price_to_item
)
```

```python
missing_dates = clean_df["transaction_date"].isnull().sum()

print("Missing transaction dates:", missing_dates)
```

```python
clean_df = clean_df.dropna(subset=["transaction_date"])
```

```python
print("\nRemaining Missing Values:")
print(clean_df.isnull().sum())

print("\nRemaining Duplicates:")
print(clean_df.duplicated().sum())
```

```python
clean_df = clean_df.dropna(subset=['price_per_unit', 'total_spent'])
```

```python
clean_df["revenue"] = (clean_df["quantity"] * clean_df["price_per_unit"])

clean_df.head()
```

```python
avg_transaction = clean_df["revenue"].mean()
print("Average Transaction Value:", avg_transaction.round(2))
```

```python
total_transactions = clean_df["transaction_id"].nunique()

total_quantity = clean_df["quantity"].sum()

total_revenue = clean_df["revenue"].sum()

average_transaction = clean_df["revenue"].mean()

average_unit_price = clean_df["price_per_unit"].mean()

print("Total Transactions:", total_transactions)
print("Total Quantity Sold:", total_quantity)
print("Total Revenue:", round(total_revenue, 2))
print("Average Transaction Value:", round(average_transaction, 2))
print("Average Unit Price:", round(average_unit_price, 2))
```

```python
item_sales = (clean_df.groupby("item").agg(
        transactions=("transaction_id", "nunique"),
        quantity_sold=("quantity", "sum"),
        revenue=("revenue", "sum")
    ) .sort_values("revenue", ascending=False))

item_sales
```

```python

clean_df["year"] = clean_df["transaction_date"].dt.year
clean_df["month"] = clean_df["transaction_date"].dt.month
clean_df["month_name"] = clean_df["transaction_date"].dt.month_name()
```

```python
monthly_sales = (
    clean_df
    .groupby(["year", "month", "month_name"])
    .agg(
        transactions=("transaction_id", "nunique"),
        revenue=("revenue", "sum")
    )
    .reset_index()
    .sort_values(["year", "month"])
)

monthly_sales
```

```python
fig = px.bar(
    item_sales.reset_index(),
    x="item",
    y="revenue",
    title="Revenue by Item",
    labels={
        "item": "Item",
        "revenue": "Revenue"
    },
    text_auto=".2s"
)

fig.update_layout(
    xaxis_title="Item",
    yaxis_title="Revenue",
    template="plotly_white"
)

fig.show()
```

```python
fig = px.bar(
    item_sales.sort_values(by="quantity_sold", ascending=False).reset_index(),
    x="item",
    y="quantity_sold",
    title="Quantity Sold by Item",
    labels={
        "item": "Item",
        "quantity_sold": "Quantity Sold"
    },
    text_auto=".2s"
)

fig.update_layout(
    xaxis_title="Item",
    yaxis_title="Quantity Sold",
    template="plotly_white"
)

fig.show()
```

```python
fig = px.line(
    monthly_sales,
    x="month_name",
    y="revenue",
    markers=True,
    title="Monthly Revenue Trend"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Revenue",
    template="plotly_white"
)

fig.show()
```

```python
clean_df.to_csv(
    "dirty_cafe_sales_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
